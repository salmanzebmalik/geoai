```python
import os
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO
MODEL_PATH = "runs/construction/v1/weights/best.pt"  
IMAGE_SOURCE = "path/to/your/satellite_images/"     
OUTPUT_DIR = "predictions/output_visualizations"

model = YOLO(MODEL_PATH)
results = model.predict(
    source=IMAGE_SOURCE,
    imgsz=640,          # to change depending on the size of the trainig images ending up using 
    conf=0.25,          
    device="cuda",       
    save=True, 
    project=OUTPUT_DIR, 
    exist_ok=True
)
```

```python
for i, result in enumerate(results):
    annotated_img = result.plot()
    annotated_img_rgb = cv2.cvtColor(annotated_img, cv2.COLOR_BGR2RGB)
    
    boxes = result.boxes
    for box in boxes:
        xyxy = box.xyxy[0].tolist() 
        conf = box.conf[0].item()
        cls_id = int(box.cls[0].item())
        cls_name = model.names[cls_id]

    plt.figure(figsize=(10, 10))
    plt.imshow(annotated_img_rgb)
    plt.axis("off")
    plt.title(f"Pred for Image {i+1}")
    plt.show()
```
